```python
def generate_text(model, tokinizer, seed_text, max_len, n_words, top_k=5):
    for _ in range(n_words):

        seq = tokinizer.texts_to_sequences([seed_text])[0]
        seq = seq[-max_len:]

        seq = pad_sequences(
            [seq],
            maxlen=max_len,
            padding="pre"
        )

        pred = model.predict(seq, verbose=0)[0]

        # Get top-k word indices
        top_indices = np.argsort(pred)[-top_k:]

        # Choose the most probable among top-k
        next_index = top_indices[-1]

        next_word = index_to_word.get(next_index, "<unknown>")

        if next_word == "<unknown>":
            break

        seed_text += " " + next_word

    return seed_text
```

```python
def generate_text(model, tokinizer, seed_text, max_len, n_words, top_k=10):

    for _ in range(n_words):

        seq = tokinizer.texts_to_sequences([seed_text])[0]
        seq = seq[-max_len:]

        seq = pad_sequences(
            [seq],
            maxlen=max_len,
            padding="pre"
        )

        predictions = model.predict(seq, verbose=0)[0]

        # Select only the top-k predictions
        top_indices = np.argsort(predictions)[-top_k:]

        # Get probabilities for top-k words
        top_probs = predictions[top_indices]

        # Normalize probabilities
        top_probs = top_probs / np.sum(top_probs)

        # Randomly choose from top-k
        next_index = np.random.choice(
            top_indices,
            p=top_probs
        )

        next_word = index_to_word.get(next_index, "<unknown>")

        if next_word == "<unknown>":
            break

        seed_text += " " + next_word

    return seed_text
```
